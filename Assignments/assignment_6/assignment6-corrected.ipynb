````python
# Write your answer here:
weight_values = pd.read_csv("data/features.csv")

check_weight = lambda weight: weight <= 3000 

#use .apply() to create a new column 'low_weight' based on the check_weight function
weight_values['low_weight'] = weight_values['weight'].apply(check_weight)

#print number of cars with weight less than or equal to 3000
num_low_weight_cars = weight_values['low_weight'].sum()
print(num_low_weight_cars)
````

(b) Subsetting Data

<font size = "3" >

- Read in the dataset named "features.csv".
- Define a global variable called ```threshold``` that takes the <br>
value 2500
- Subset the data to get one dataset with weight $>$ threshold and <br>
another with weight $\le$ threshold. Use the ```.query()``` function.
- Compute the mean acceleration for each of these datasets, printing these values to the screen<br>

Note: Review how to define a global variable inside ```.query()```
 

````python
# Write your answer here:
weight_values = pd.read_csv("data/features.csv")
threshold = 2500 

high_weight_cars = weight_values.query("weight > @threshold")
low_weight_cars = weight_values.query("weight <= @threshold")

print(high_weight_cars['acceleration'].mean())
print(low_weight_cars['acceleration'].mean())
````

````python
# Write your answer here:
fifa_players = pd.read_csv("data/fifa23_players_basic.csv")

#sort values by age
sorted_fifa = fifa_players.sort_values(by='Age', ascending=False)

#subset of 5 oldest players
oldest_five = sorted_fifa.iloc[:5]
print(oldest_five)
````

(d) Use queries + global variables

<font size = "3">

- Read the "fifa23_players_basic.csv" dataset
- Compute the mean of the column "Weight(in kg)"
and store it as a variable
- Subset the players whose Weight in kilograms is stricly large than this mean value
  using ".query()"


HINT: Remember that there is a special syntax for when using column names inside the .query() method (the backtick `)

````python
# Write your answer here:
fifa23_players = pd.read_csv("data/fifa23_players_basic.csv")

avg_weight = fifa23_players["Weight(in kg)"].mean()

over_avg_weight = fifa23_players.query("`Weight(in kg)` > @avg_weight")

print(over_avg_weight)
````

Read in the dataset named "wdi_urban.csv".

  This contains data from 1980 and 2020.

- **Create an array with the unique values of "year" using the ``Pandas`` function ``unique``**

- **Create a plot with a ...**
    - Histogram of "prop_urbanpopulation" (proportion of urban population) for each year
    - Set the histogram option `alpha = 0.5` for transparency
    - All the histograms should be overlapping on the same plot
    - Create a legend with the list of years
    - Label the axes and the title

**Hint**: The procedure will be similar (but not exactly the same!) as when we created a scatter plot of cars with different numbers of cylinders in Lecture 11

````python
# your answer here

df = pd.read_csv("data/wdi_urban.csv")

#creat an array w unique year using uinque()
unique_years = df['year'].unique()

#plot a histogram of the unique years
for year in unique_years:
    year_data = df.query("year == @year")
    plt.hist(year_data['prop_urbanpopulation'], alpha=0.5, label=str(year))

plt.xlabel("Urban Population")
plt.ylabel("Frequency")
plt.legend(labels=unique_years, title="Year")
plt.title("Histogram of Urban Population by Year")
plt.show()

    

````
