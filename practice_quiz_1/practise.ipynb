```python
import numpy as np
import matplotlib.pyplot as plt

x_vals = [1, 2, 3, 4, 5]
y_vals = [2, 1, 0, -1, -2]
```

```python
z = 2*np.array(x_vals) + np.array(y_vals)**2

plt.hist(z, bins= 20,color="red")
plt.xlabel(" z values ")
plt.ylabel ( "frequency ")
plt.title( "histogram of equation z")
```

```python
np.random.seed(0)
n = 1000

fig, subfig_array = plt.subplots(nrows=1, ncols = 2, figsize = (6,3)) 

normal_samples = np.random.normal(loc=0, scale=1, size = n)
exponentional_samples = np.random.exponential(scale = 1, size =n)

#first sub figure 
subfig_array[0].hist(x=normal_samples)
subfig_array[0].set_title("Normal Distribution (1000 samples)")
subfig_array[0].set_xlabel("sample value")
subfig_array[0].set_ylabel("frequency")

#second figure 
subfig_array[1].hist(x=exponentional_samples)
subfig_array[1].set_title("Exponentional (1000 samples)")
subfig_array[1].set_xlabel("sample value")
subfig_array[1].set_ylabel("frequency")

plt.tight_layout()

plt.show()

```

```python
#chatgpt 
np.random.seed(0)
n = 1000
a = np.random.normal(loc=0, scale=1, size=n)
b = np.random.exponential(scale=1, size=n)

fig, axes = plt.subplots(1, 2, figsize=(10,4), sharey=True)

axes[0].hist(a, bins=30)
axes[0].set_title("Normal(0,1)")
axes[0].set_xlabel("Value")
axes[0].set_ylabel("Frequency")

axes[1].hist(b, bins=30)
axes[1].set_title("Exponential(1)")
axes[1].set_xlabel("Value")
axes[1].set_ylabel("Frequency")

plt.tight_layout()
plt.show()
```
