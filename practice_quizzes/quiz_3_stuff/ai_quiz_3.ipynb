## Question 2: Filter → Group → Aggregate → Filter (with Chaining)

**Instructions:** You will extract a subset of the data based on a logical condition, then group this subset and aggregate statistics. Finally, extract rows where one of the aggregated statistics is greater than or equal to some value. **Do this in 1 line using method chaining for full credit.**

**Dataset:** Create this DataFrame:

```python
sales_df = pd.DataFrame({
    'region': ['North', 'South', 'North', 'East', 'West', 'South', 'North', 'East', 'West', 'South', 'North', 'East'],
    'product': ['A', 'B', 'A', 'C', 'B', 'A', 'C', 'B', 'A', 'C', 'B', 'A'],
    'sales': [1200, 800, 1500, 600, 950, 1100, 700, 1300, 1400, 850, 1600, 1250],
    'units': [30, 20, 35, 15, 25, 28, 18, 32, 33, 22, 38, 31]
})
```

**Your Task:**
- Filter for rows where `sales >= 1000`
- Group by `region`
- Aggregate: compute the mean of `sales` and the sum of `units`
- Filter the result to only show regions where the sum of `units` is >= 60
- **Do all of this in ONE line using method chaining**

**Hint:** Use `.query()`, `.groupby()`, `.agg()`, and `.query()` again (or boolean indexing)


````python
sales_df = pd.DataFrame({
    'region': ['North', 'South', 'North', 'East', 'West', 'South', 'North', 'East', 'West', 'South', 'North', 'East'],
    'product': ['A', 'B', 'A', 'C', 'B', 'A', 'C', 'B', 'A', 'C', 'B', 'A'],
    'sales': [1200, 800, 1500, 600, 950, 1100, 700, 1300, 1400, 850, 1600, 1250],
    'units': [30, 20, 35, 15, 25, 28, 18, 32, 33, 22, 38, 31]
})
````

````python


sales_agg = (sales_df.query("sales >= 1000")
             .groupby(["region"])
             .agg(mean_sales = ('sales', 'mean'),
                  units_sum = ('units', 'sum'))
             .query("units_sum >= 60"))

print(sales_agg)

````

## Question 3: Recode Numeric to Categorical

**Instructions:** You have a DataFrame with a numeric column called 'temperature'. Recode this numeric variable into a categorical variable (string values) based on the following intervals:
- If temperature < 50: "Cold"
- If 50 <= temperature < 75: "Moderate"
- If temperature >= 75: "Hot"

**Dataset:** Create this DataFrame:

```python
weather_df = pd.DataFrame({
    'city': ['Atlanta', 'Boston', 'Chicago', 'Denver', 'Miami', 'Phoenix', 'Seattle', 'Austin'],
    'temperature': [68, 45, 52, 48, 88, 95, 55, 78]
})
```

**Your Task:**
- Create a new column called 'temp_category' that contains the appropriate string based on the temperature intervals
- Display the DataFrame with the new column

**Hint:** You can use `.apply()` with a function, `np.select()`, or `pd.cut()` with custom labels


````python
weather_df = pd.DataFrame({
    'city': ['Atlanta', 'Boston', 'Chicago', 'Denver', 'Miami', 'Phoenix', 'Seattle', 'Austin'],
    'temperature': [68, 45, 52, 48, 88, 95, 55, 78]
})
````

````python
def categorize_temp(temp):
    if temp < 50:
        return 'Cold'
    elif temp < 75:
        return 'Moderate'
    else:
        return 'Hot'

weather_df['temp_category'] = weather_df['temperature'].apply(categorize_temp)

#this method used .apply() with a custom function to create a new column based on conditions
````

## Question 4: Group → Aggregate → Sort (with Chaining)

**Instructions:** Group a dataset by one of its columns, aggregate one or more statistics for each group, then sort the data by one of the aggregated statistics in **descending order**. **Do this in 1 line using method chaining for full credit.**

**Dataset:** Create this DataFrame:

```python
employees_df = pd.DataFrame({
    'department': ['HR', 'IT', 'Sales', 'HR', 'IT', 'Sales', 'HR', 'IT', 'Sales', 'IT'],
    'employee': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve', 'Frank', 'Grace', 'Henry', 'Ivy', 'Jack'],
    'salary': [55000, 75000, 62000, 58000, 80000, 68000, 52000, 72000, 71000, 77000],
    'years_experience': [3, 5, 4, 4, 7, 5, 2, 6, 6, 5]
})
```

**Your Task:**
- Group by `department`
- Aggregate: compute the mean of `salary` and the max of `years_experience`
- Sort by mean salary in **descending order**
- **Do all of this in ONE line using method chaining**

---


````python
employees_df = pd.DataFrame({
    'department': ['HR', 'IT', 'Sales', 'HR', 'IT', 'Sales', 'HR', 'IT', 'Sales', 'IT'],
    'employee': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve', 'Frank', 'Grace', 'Henry', 'Ivy', 'Jack'],
    'salary': [55000, 75000, 62000, 58000, 80000, 68000, 52000, 72000, 71000, 77000],
    'years_experience': [3, 5, 4, 4, 7, 5, 2, 6, 6, 5]
})
````

````python

employees_agg = (employees_df.groupby("department")
                 .agg(avg_salary = ('salary', 'mean'),
                      max_experience = ('years_experience', 'max'))
                 .sort_values("avg_salary", ascending=False))

#result = employees_df.groupby('department').agg({'salary': 'mean', 'years_experience': 'max'}).sort_values('salary', ascending=False)
````

## Question 6: Merge Two DataFrames

**Instructions:** Merge two DataFrames. After merging, the primary DataFrame will have one new column that is obtained from the secondary DataFrame.

**Dataset:** Create these two DataFrames:

```python
# Primary DataFrame
products_df = pd.DataFrame({
    'product_id': [101, 102, 103, 104, 105],
    'product_name': ['Laptop', 'Mouse', 'Keyboard', 'Monitor', 'Webcam'],
    'price': [899, 25, 75, 299, 89],
    'category_id': [1, 2, 2, 1, 3]
})

# Secondary DataFrame
categories_df = pd.DataFrame({
    'category_id': [1, 2, 3, 4],
    'category_name': ['Electronics', 'Accessories', 'Peripherals', 'Software']
})
```

**Your Task:**
- Merge `products_df` with `categories_df` so that `products_df` gains the `category_name` column
- Use the appropriate merge/join method
- Display the merged DataFrame

**Hint:** Use `.merge()` and specify the key column to merge on

---

````python
products_df = pd.DataFrame({
    'product_id': [101, 102, 103, 104, 105],
    'product_name': ['Laptop', 'Mouse', 'Keyboard', 'Monitor', 'Webcam'],
    'price': [899, 25, 75, 299, 89],
    'category_id': [1, 2, 2, 1, 3]
})

# Secondary DataFrame
categories_df = pd.DataFrame({
    'category_id': [1, 2, 3, 4],
    'category_name': ['Electronics', 'Accessories', 'Peripherals', 'Software']
})
````

````python
products_df = pd.merge(left = products_df, 
                            right = categories_df, 
                            on = 'category_id', 
                            how = 'left')
````
