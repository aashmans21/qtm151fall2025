```python
#q1 

# --- Code Cell 1 (Given, correct) ---
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-5, 5, 50)
y = 2 * x**2 + 3 * x + 1 + np.random.randn(50) * 5

# Fit quadratic model
coeffs = np.polyfit(x, y, 2)
a, b, c = coeffs
```

```python
# --- Code Cell 2 (Needs correction) ---
y_pred = a*x**2 + b*x + c
plt.scatter(x, y)
plt.plot(x, y_pred)
plt.show()
```
