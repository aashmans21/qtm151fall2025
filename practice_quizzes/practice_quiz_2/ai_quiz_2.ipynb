## Question 1: Function Error Debugging
**Instructions:** The code cell below defines a function with one input argument (a number) and one return value (a string). Then a variable is defined, passed to the function, and the output is printed. However, there is a mistake in the code and an error occurs. Add a comment to the offending line explaining what happened, then run a corrected version in another cell.

````python
#error cell 

def classify_temperature(temp):
    if temp > 75:
        return "Hot"
    elif temp > 50:
        return "Moderate"
    else:
        return "Cold"

my_temp = "68" # string instead of number
result = classify_temperature(my_temp)
print(result)
````

````python
#corrected code

def classify_temperature(temp):
    if temp > 75:
        return "Hot"
    elif temp > 50:
        return "Moderate"
    else:
        return "Cold"

my_temp = "68" #or just use a number directly
result = classify_temperature(int(my_temp))
print(result)
````

````python
def label_num(x):
    if x > 0:
        msg = "positive"
    # BUG: for x <= 0 we never define msg, so returning it raises UnboundLocalError
    return msg

n = -3
print(label_num(n))

# msg is only defined in the if-branch; for x <= 0 the function returns an undefined variable -> UnboundLocalError
````

````python
def label_num(x):
    if x > 0:
        msg = "positive"
    else:
        msg = "non-positive"
    return msg

n = -3
print(label_num(n))
````

mixing types: "age: " + 30 (needs str(30) or an f-string)

missing return

using = instead of == in conditions

function defined after it’s called (call must come after definition or in another cell run later)

## Question 2: NumPy Array Operation Issue
**Instructions:** The code below defines two lists of numbers, then attempts to create a NumPy array from the entries of the two lists using a mathematical operation. The code runs fine, but the output is different than intended. Add a comment explaining why the output is doing something differently, then run a corrected version in another cell.

````python
import numpy as np

list1 = [1, 2, 3, 4, 5]
list2 = [10, 20, 30, 40, 50]

# The intention is to create an array where each element is the sum of corresponding elements
result = np.array(list1 + list2) # BUG: this concatenates the lists instead of summing element-wise
print(result)
````

````python
#corrected code
import numpy as np

list1 = [1, 2, 3, 4, 5]
list2 = [10, 20, 30, 40, 50]

# The intention is to create an array where each element is the sum of corresponding elements
result = np.array(list1) + np.array(list2)
print(result)
````

## Question 3: DataFrame Filtering
**Instructions:** Create a DataFrame by reading in the provided .csv file. Extract the subset of data where the 'age' column is greater than or equal to 25. Save this extracted subset to a separate DataFrame. Then print out the number of rows in the extracted dataset.

**Dataset:** `students.csv`

**Your Task:**
- Read the CSV file into a DataFrame
- Filter for rows where age >= 25
- Save to a new DataFrame
- Print the number of rows that satisfied the condition

````python
import pandas as pd
df = pd.read_csv("data_folder/students.csv")

# Filter for age >= 25
filtered_df = df[df['age'] >= 25]

# Print the number of rows
num_rows = len(filtered_df)
print(f"Number of students aged 25 or older: {num_rows}")
# Alternative: print(filtered_df.shape[0])
````

## Question 4: Custom Function with .apply()
**Instructions:** Write a function that takes in a single number (representing a grade percentage) as an input argument and returns a single string as output. The function should return "Pass" if the grade is 60 or above, and "Fail" if it's below 60. Then use the `.apply()` method to evaluate this function for every row in the DataFrame, using the 'test_score' column as input.

**Dataset:** Create this DataFrame first:
```python
import pandas as pd

grades_df = pd.DataFrame({
    'student_id': [101, 102, 103, 104, 105, 106],
    'test_score': [85, 52, 73, 45, 90, 67]
})
```

**Your Task:**
- Write a function that takes a number and returns "Pass" or "Fail"
- Use `.apply()` to apply this function to the 'test_score' column
- Add the results as a new column called 'status'

````python
import pandas as pd

grades_df = pd.DataFrame({
    'student_id': [101, 102, 103, 104, 105, 106],
    'test_score': [85, 52, 73, 45, 90, 67]
})
````

````python
#define the function
def pass_fail(score):
    if score >= 60:
        return "Pass"
    else:
        return "Fail"
    
#apply the function 

grades_df['status'] = grades_df['test_score'].apply(pass_fail)

print(grades_df)
````
